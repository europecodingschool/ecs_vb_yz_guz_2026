```python
# While Loop - While Döngüsü

#Birikim hedefi

hedef = 10000
aylik = 1750
birikim = 0
ay = 0

while birikim < hedef:
  ay += 1 # her döngüde aya 1 ekle. ay = ay + 1
  birikim += aylik # birikim = birikim + aylik
  print(ay,birikim)

```

```python
# Basit bir tahmin oyunu.
# 1. Tahmin etmek istediğimiz sayı olsun. "gizli"
# 2. 3 tahmin hakkımız olsun.
# 3. input ile tahmin alalım.
# 4. bildiyse bildin.
# 5. input küçükse daha büyük bir sayı gir uyarısı.
# 6. input büyükse daha küçük bir sayı gir uyarısı.
# 7. hakkı biterse de hakkın bitti.
```

```python
gizli = 50 #1. adım
tahmin_hakki = 3 #2. adım

while tahmin_hakki > 0:
  tahmin = int(input("Tahmininizi girin:")) #3. adım

  if tahmin == gizli:
    print("Tebrikler. Bildiniz.")
    break
  elif tahmin < gizli:
    print("Daha büyük bir sayı giriniz.")
  else:
    print("Daha küçük bir sayı giriniz.")

  tahmin_hakki -= 1 # tahmin_hakki = tahmin_hakki - 1

if tahmin_hakki == 0:
  print("Hakkınız bitti.gizli sayi:",gizli)
```
